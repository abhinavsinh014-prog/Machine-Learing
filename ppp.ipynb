```python
from ast import Expression
import numpy as np
```

```python
def input_matrix(name):
    print(f"Enter the number of rows and columns of the {name} matrix:")
    try:
        return np.array(eval(input()))
    except Exception as e:
        print("Invalid input. Please enter a valid matrix.")
        return input_matrix(name)
    
def input_vector(name):
    print(f"Enter the elements of the {name} vector:")
    try:
        return np.array(eval(input()))
    except Exception as e:
        print("Invalid input. Please enter a valid vector.")
        return input_vector(name)

def matrix_multiply(A,B):
    try:
        return np.dot(A,B)
    except Exception as e:
        return f"ArithmeticError: {e}. Please check the dimensions of the matrices."
       
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum()

def tanh(x):
    return np.tanh(x)

while True:
    print("Choose an operation:")
    print("1. Matrix Multiplication")
    print("2. vector multiplication")
    print("3. Sigmoid Activation Function")
    print("4. ReLU Activation Function")
    print("5. Softmax Activation Function")
    print("6. Tanh Activation Function")
    print("7. Exit")

    choice = input("Enter your choice (1-7): ")
    if choice == '1':
        A = input_matrix("first")
        B = input_matrix("second")
        result = matrix_multiply(A, B)
        print("Result of Matrix Multiplication:")
        print(result)
    elif choice == '2':
        A = input_vector("first")
        B = input_vector("second")
        result = matrix_multiply(A, B)
        print("Result of Vector Multiplication:")
        print(result)
    elif choice == '3':
        x = input_vector("input")
        result = sigmoid(x)
        print("Result of Sigmoid Activation Function:")
        print(result)
    elif choice == '4':
        x = input_vector("input")
        result = relu(x)
        print("Result of ReLU Activation Function:")
        print(result)
    elif choice == '5':
        x = input_vector("input")
        result = softmax(x)
        print("Result of Softmax Activation Function:")
        print(result)
    elif choice == '6':
        x = input_vector("input")
        result = tanh(x)
        print("Result of Tanh Activation Function:")

        print(result)
    elif choice == '7':
        break
    else:
        print("Invalid choice. Please try again.")
```
