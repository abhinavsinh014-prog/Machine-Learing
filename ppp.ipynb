```python
import numpy as np
```

```python
def input_matrix(name):
    print(f"Enter the number of rows and columns of the {name} matrix:")
    try:
        return np.array(eval(input()))
    except Exception as e:
        print("Invalid input. Please enter a valid matrix.")
        return input_matrix(name)

def matrix_multiply(A,B):
    try:
        return np.dot(A,B)
    except Exception as e:
        return f"ArithmeticError: {e}. Please check the dimensions of the matrices."
    
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum()
```
