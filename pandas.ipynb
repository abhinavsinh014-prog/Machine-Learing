```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'a': [1, 2, 3,np.nan, 5],
    'b': [5, 6, np.nan, 8, 9],
    'c': [10, 11, 12, 13, 14],
    'd': [15, None, None, None, 19]
})
```

```python
print(df.isna().sum())
```

```python
clean_df = df.dropna()  
print(clean_df)
```

```python
clean_df = df.dropna(how='all')  
print(clean_df)
```

```python
df_there = df.fillna(0)
print(df_there)
```

```python
dfo = df.dropna(axis=1, thresh=3)
print(dfo)
```
